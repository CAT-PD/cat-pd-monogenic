## Define paths for input and out files

```python
# CRAM input directory
cram_directory = Path('/path/to/home')

# CRAM manifest
manifest = f'{cram_directory}/sample_manifest.tsv'

# Genome (hg38)
genome = 38

# Reference (hg38)
reference = '/path/to/reference/fasta/hg38/Homo_sapiens_assembly38.fasta'

# Output directory
output_dir = Path('/path/to/home/Gauchian_output')
output_dir.mkdir(exist_ok=True)
```

## Get list of .cram absolute paths

```python
# Get absolute paths for all .cram files
cramList = [str(f.resolve())
            for f in cram_directory.glob('*_Cram/**/*.cram')
            if f.is_file()]
```

```python
# Convert to dataframe and save as manifest
df = pd.DataFrame({"cram_path": cramList})
df.to_csv(manifest, index=False, header=False)
```

```python
print('Number of samples: ', len(cramList))
```

## Run Gauchian

```python
# Define function to run Gauchian
out_name = 'gauchian_output'
threads = 16

# Define the process to run Gauchian
run_gauchian = [
                'gauchian',
                '--manifest', str(manifest),
                '--genome', str(genome),
                '--reference', str(reference),
                '--threads', str(threads),
                '--outDir', str(output_dir),
                '--prefix', str(out_name),
                ]
# Run per each sample
try:
    subprocess.run(run_gauchian, check=True)
# Print error and continue
except Exception as e:
    print('Error: ', e)
```
